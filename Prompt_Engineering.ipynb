### Few-Shot Prompting

```python
# sentiment classification

client = OpenAI()

prompt = """
  classifiy the sentiment.

  Review: The product exceeded my expectations.
  Sentiment: Positive

  Review: The delivery was late and the item was damaged.
  Sentiment: Negative

  Review: The movie was okay.
  Sentiment: Neutral

  Review: The food was delicious and arrived quickly.
  Sentiment:
"""

response = client.responses.create(
    model='gpt-4o-mini',
    input=prompt
)

print(response.output_text)
```

### Chain-of-Thought Prompting

```python
client = OpenAI()

prompt = """
  Question:
  A box contains 10 balls.
  3 are removed.

  Reasoning: 10 - 3 = 7

  Answer: 7

  Question:
  A shop has 20 apples.
  It sells 8.

  Reasoning: 20 - 8 = 12

  Answer: 12

  Question:
  A train has 80 passengers.
  15 leave.

  Reasoning:
"""

response = client.responses.create(
    model='gpt-4o-mini',
    input=prompt
)

print(response.output_text)
```

### System Prompts

##### Customer Support Bot

```python
client = OpenAI()

response = client.responses.create(
    model='gpt-4o-mini',
    instructions="""
      You are a customer support assistant.

      Rules:
      - Be Polite.
      - If uncertain, admit uncertainty.
      - Keep answers under 120 words.
    """,
    input="Can I get a refund after 45 days?"
)

print(response.output_text)
```

##### JSON Assistant

```python
client = OpenAI()

response = client.responses.create(
    model="gpt-4o-mini",
    instructions="""
      Always respond using JSON.

      Schema:
      {
        "answer": "",
        "confidence": ""
      }
    """,
    input="What is Redis?"
)

print(response.output_text)
```
