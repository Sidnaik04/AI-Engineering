### Structured Output

##### Simple JSON Output:

```python
import json

client = OpenAI()

prompt = """
  Extract the following information.

  John Smith works at Google in California.

  Return JSON only.

  {
      "name":"",
      "company":"",
      "location":""
  }
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}],
    response_format={"type": "json_object"}
)

data = json.loads(response.choices[0].message.content)

print(data)
```

##### Production-Ready Structured Outputs

```python
from pydantic import BaseModel

client = OpenAI()

class Resume(BaseModel):
  name: str
  email:str
  skills: list[str]

response = client.responses.parse(
    model="gpt-4o-mini",
    input="""
      Extract information.

      John Smith
      john@gmail.com

      Skills:
      Python
      FastAPI
      PostgreSQL
    """,
    text_format=Resume,
)

resume = response.output_parsed

print(resume.name)
print(resume.skills)
```

### Evaluation of Prompt

```python
test_cases = [
    {
        "email": "Meeting at 4 PM tomorrow.",
        "expected": "work"
    },
    {
        "email": "You won an iPhone!",
        "expected": "spam"
    },
    {
        "email": "Amazon shipped your order.",
        "expected": "promotion"
    },
    {
        "email": "Let's catch up this weekend.",
        "expected": "personal"
    }
]
```

```python
client = OpenAI()

prompt = """
  Clasify the email.

  Return only one category:
  - work
  - personal
  - promotion
  - spam
"""

correct=0

for sample in test_cases:

  response = client.responses.create(
      model='gpt-4o-mini',
      input=f"""
        {prompt}

        Email:
        {sample['email']}
      """
  )

  prediction = response.output_text.strip().lower()

  if prediction == sample['expected']:
    correct += 1

accuracy = correct / len(test_cases)

print(f"Accuracy: {accuracy:.2%}")
```
