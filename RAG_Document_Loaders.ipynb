#### Eager `load()` vs. Memory-Efficient `lazy_load()`:

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader('/content/python-crash-course.pdf')

# --- Approach 1: Eager Load (Loads all pages into RAM at once) ---
docs = loader.load()
print(f"Loaded {len(docs)} pages in memory.")
```

```python
# --- Approach 2: Lazy Load (Streams pages one-by-one) ---
for doc in loader.lazy_load():
    # Process or index each page individually without high RAM overhead
    print(f"Processing page {doc.metadata['page']}: {len(doc.page_content)} characters")
```

#### Async Web Page Loading

```python
import asyncio
from langchain_community.document_loaders import WebBaseLoader

async def fetch_urls():
  urls = [
      "https://en.wikipedia.org/wiki/Cosine_similarity",
      "https://en.wikipedia.org/wiki/Dot_product",
      "https://en.wikipedia.org/wiki/Euclidean_distance"
  ]

  loader = WebBaseLoader(urls)

  # Stream documents concurrently without blocking the event loop
  async for doc in loader.alazy_load():
      print(f"Fetched: {doc.metadata['source']} | Length: {len(doc.page_content)}")

await fetch_urls()
```

```python
from marker.converters.pdf import PdfConverter
from marker.models import create_model_dict
from marker.output import text_from_rendered

# 1. Initialize the converter with required model dependencies
converter = PdfConverter(artifact_dict=create_model_dict())

# 2. Convert the PDF file(returns a Pydantic object)
rendered = converter('AI Engineering.pdf')

# 3. Extract text, metadata and extracted images
markdown_text, metadata, images = text_from_rendered(rendered)

print(markdown_text)
```

```python
from google.colab import userdata
from mistralai.client import Mistral

api_key = userdata.get('MISTRAL_API_KEY')
model = "mistral-medium-latest"

client = Mistral(api_key=api_key)

chat_response = client.chat.complete(
    model = model,
    messages = [
        {
            "role": "user",
            "content": "What is the best French cheese?",
        },
    ]
)

chat_response
```

```python
from google.colab import userdata
from mistralai.client import Mistral

api_key = userdata.get('MISTRAL_API_KEY')
client = Mistral(api_key=api_key)

with open("AI Engineering.pdf", "rb") as file:
    uploaded_file = client.files.upload(
        file={"file_name": "AI Engineering.pdf", "content": file.read()},
        purpose="ocr"
    )

signed_url = client.files.get_signed_url(file_id=uploaded_file.id, expiry=1)

pdf_response = client.ocr.process(
    document={"document_url": signed_url.url},
    model="mistral-ocr-latest"
)

# Extract and combine the markdown blocks page by page
markdown_text = "\n\n".join([page.markdown for page in pdf_response.pages[:5]])
print(markdown_text) # first 5 pages
```
